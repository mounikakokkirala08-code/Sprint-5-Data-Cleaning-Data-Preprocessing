# 12 Business Insights


## Step 1 – Understand the Concept (Markdown)


```python
Business Insights from EDA

Business Insights means converting the results of Exploratory Data Analysis (EDA) into meaningful observations that can help in business decision-making.

EDA tells us what is happening in the data, while business insights explain why it matters and what action could be considered.

The process is:

Visualization → Observation → Insight → Business Recommendation

1. Important Patterns

Patterns are repeated behaviors or characteristics found in the data.

Example:

Larger properties generally have higher house prices.

2. Trends

Trends show how one variable changes as another variable increases or decreases.

Example:

House prices generally increase as property area increases.

3. Relationships

Relationships show how two or more variables are connected.

Example:

Furnished properties tend to have higher average prices than unfurnished properties.

4. Anomalies

Anomalies are unusual observations that differ significantly from the general pattern.

Example:

A few properties have unusually high prices compared with properties having similar areas.

5. Potential Business Problems

EDA can reveal problems that may affect business decisions.

Examples:

Highly unequal property prices
Rare property categories
Missing information
Extreme price outliers
Strongly concentrated locations
Limited data for certain property segments
6. Interesting Segments

Segments are groups of properties that have similar characteristics.

Examples:

High-priced properties
Budget properties
Large-area properties
Furnished properties
Urban properties
Villas
7. Potential Opportunities

Insights can identify areas where a business could improve or grow.

Example:

If furnished properties consistently command higher prices, sellers could consider furnishing properties before listing them, provided the additional investment is justified by the expected price premium.
```

## Step 2 – Demonstrate the Concept (Markdown)

Instead of writing:

"The graph shows average prices by property type."

We should write:

Observation: Villas have the highest average price, followed by independent houses and apartments.

Insight: Property type appears to be an important factor associated with house prices, with villas commanding a substantially higher average price than apartments.

Business Opportunity: Real-estate businesses could focus premium-property marketing on villas while positioning apartments toward budget-conscious buyers.

## Step 3 – Implement the Concept (Python)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("house_price_prediction_cleaned.csv")

# Target variable
target = "Price_lakhs"

print("Dataset Shape:", df.shape)
display(df.head())
```

```python
#Check Important Patterns
df.describe()
df["Property_Type"].value_counts()
```

```python
#4. Check Average Price by Property Type
df.groupby("Property_Type")["Price_lakhs"].mean()
```

```python
# 5. Check Average Price by Locality
df.groupby("Locality_Type")["Price_lakhs"].mean()
```

```python
#6. Check Area and Price Relationship
df[["Area_sqft", "Price_lakhs"]].corr()
```

```python
#7. Visualize Area vs Price
sns.scatterplot(
    data=df,
    x="Area_sqft",
    y="Price_lakhs"
)


plt.title("Area vs Price")
plt.show()
```

```python
#8. Compare Property Type and Price
sns.boxplot(
    data=df,
    x="Property_Type",
    y="Price_lakhs"
)


plt.title("Property Type vs Price")
plt.xticks(rotation=30)
plt.show()
```

```python
#9. Find Highest-Priced Property
df.loc[df["Price_lakhs"].idxmax()]
```

```python
#10. Find Lowest-Priced Property
df.loc[df["Price_lakhs"].idxmin()]
```

```python
#11. Find Price Outliers
sns.boxplot(x=df["Price_lakhs"])

plt.title("Price Outliers")
plt.show()
```

# Duplicate Data Handling

```python
## What are Duplicate Records?
### Step 1 – Understand the Concept (Markdown)
Definition: Duplicate records are identical or repeated rows that appear more than once in a dataset.

In simple words: Duplicate Record = Same data repeated multiple times.

Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Name	Age	Salary
Mani	25	30000
Roja	30	40000
Mani	25	30000
Mahi	35	50000

### Step 3 – Implement the Concept (Python)

```

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Mani", "Roja", "Mani", "Mahi"],
    "Age": [25, 30, 25, 35],
    "Salary": [30000, 40000, 30000, 50000]
})

# Detect duplicate rows
print(df.duplicated())
```

## Partial Duplicates
### Step 1 – Understand the Concept (Markdown)
Definition: Partial duplicates are records where some important column values are the same, but the entire row is not identical.

In simple words: Partial Duplicate = Some values are repeated, but the complete row is different.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Name	Age	City
Mani	25	Hyderabad
Mani	25	Chennai
Roja	30	Delhi
    
### Step 3 – Implement the Concept (Python)




```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Mani", "Mani", "Roja"],
    "Age": [25, 25, 30],
    "City": ["Hyderabad", "Chennai", "Delhi"]
})

# Find partial duplicates based on Name and Age
print(df.duplicated(subset=["Name", "Age"]))
```

## Exact Duplicates
### Step 1 – Understand the Concept (Markdown)
Definition: Exact duplicates are records where all column values are exactly the same as another record.

In simple words: Exact Duplicate = The entire row is repeated.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Name	Age	Salary
Mani	25	30000
Roja	30	40000
Mani	25	30000

The first and third rows are exact duplicates because every value is identical.

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Mani", "Roja", "Mani"],
    "Age": [25, 30, 25],
    "Salary": [30000, 40000, 30000]
})

# Detect exact duplicates
print(df.duplicated())
```

```python
## Identifying Duplicates
### Step 1 – Understand the Concept (Markdown)
Definition: Identifying duplicates means finding repeated records in a dataset before removing or handling them.

In simple words: Identifying Duplicates = Finding repeated rows or values.

### Step 2 – Demonstrate the Concept (Markdown)
| Name | Age | Salary |
| ---- | --: | -----: |
| Mani |  25 |  30000 |
| Roja |  30 |  40000 |
| Mani |  25 |  30000 |

### Step 3 – Implement the Concept (Python)
```

```python
# Identify duplicate rows
print(df.duplicated())
```

## duplicated()
### Step 1 – Understand the Concept (Markdown)
Definition: duplicated() is a Pandas method used to identify duplicate rows in a DataFrame.

In simple words: duplicated() = Check whether a row is repeated.

By default, it marks the first occurrence as False and repeated occurrences as True.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Name	Age	Salary
Mani	25	30000
Roja	30	40000
Mani	25	30000

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Mani", "Roja", "Mani"],
    "Age": [25, 30, 25],
    "Salary": [30000, 40000, 30000]
})

# Identify duplicate rows
print(df.duplicated())

# Count duplicate rows
print(df.duplicated().sum())

# Display duplicate rows
print(df[df.duplicated()])
```

```python
## drop_duplicates()
### Step 1 – Understand the Concept (Markdown)
Definition:drop_duplicates() is a Pandas method used to remove duplicate rows from a DataFrame.

In simple words: drop_duplicates() = Remove repeated rows.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Before:

Name	Age	Salary
Mani	25	30000
Roja	30	40000
Mani	25	30000

After drop_duplicates():

Name	Age	Salary
Mani	25	30000
Roja	30	40000
```

```python
df = df.drop_duplicates()

print(df)
```

## Duplicate Detection Based on Selected Columns
### Step 1 – Understand the Concept (Markdown)
Definition: Duplicate detection based on selected columns means checking for duplicate records using only specific columns, instead of comparing the entire row.

In simple words: Selected Columns = Check duplicates using only the columns that matter.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Name	Age	City
Mani	25	Hyderabad
Mani	25	Chennai
Roja	30	Delhi

### Step 3 – Implement the Concept (Python)


```python
### Step 3 – Implement the Concept (Python)
# Detect duplicates:
import pandas as pd


df = pd.DataFrame({
    "Name": ["Mani", "Mani", "Roja"],
    "Age": [25, 25, 30],
    "City": ["Hyderabad", "Chennai", "Delhi"]
})


# Detect duplicates based on selected columns
print(df.duplicated(subset=["Name", "Age"]))
```

```python
| Topic                   | Meaning                                 |
| ----------------------- | --------------------------------------- |
| **Exact Duplicate**     | Complete row is repeated                |
| **Partial Duplicate**   | Selected columns are repeated           |
| **`duplicated()`**      | Finds duplicates                        |
| **`drop_duplicates()`** | Removes duplicates                      |
| **`subset`**            | Selects columns for duplicate checking  |
| **`keep="first"`**      | Keeps first record                      |
| **`keep="last"`**       | Keeps last record                       |
| **`keep=False`**        | Marks/removes all duplicate occurrences |
| **Business Rule**       | Decides which record should be retained |
```

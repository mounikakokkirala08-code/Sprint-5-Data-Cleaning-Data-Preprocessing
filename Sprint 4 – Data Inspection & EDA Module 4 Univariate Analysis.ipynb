# Univariate Analysis

### Step 1 – Understand the Concept
Own Words; Univariate Analysis means analyzing one variable/column at a time.

For your house-price dataset, we can analyze; Area_sqft, Bedrooms,Bathrooms
Age_years,,Location_Score, Parking_Spaces, Price_lakhs

We don't compare two variables yet.

Definition; Univariate analysis is the statistical examination of a single variable to understand its:

Center
Spread
Distribution
Shape
Variability
### Step 2 – Real-World Analogy

Suppose we want to understand only house prices.

We ask:

What is the average price?
What is the middle price?
What is the cheapest?
What is the most expensive?
How spread out are the prices?
Are most prices concentrated around the average?
Is the distribution skewed?

We are analyzing only:

Price_lakhs

That's univariate analysis.

### Visualizations

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize= (15,5))

plt.hist(df["Price_lakhs"], bins=20, edgecolor="black")
plt.title("Distribution of House Prices")
plt.xlabel("price in lakhs")
plt.ylabel("frequency")

                     
```

```python
plt.figure(figsize= (15,5))

sns.histplot(
    data=df,
    x="Price_lakhs",
    bins=10
)

plt.title("Distribution of House Prices")
plt.xlabel("price in lakhs")
plt.ylabel("frequency")
         
```

```python
for column in numeric_columns:
    
    plt.figure(figsize=(8, 5))
    
    sns.histplot(
        data=df,
        x=column,
        bins=10
    )
    
    plt.title(f"Distribution of {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    
    plt.show()
```

## KDE or Density plot

```python
plt.figure(figsize=(8, 5))

sns.kdeplot(
    data=df,
    x="Price_lakhs",
    fill=True
)

plt.title("Density Plot of House Prices")
plt.xlabel("Price (Lakhs)")
plt.ylabel("Density")

plt.show()
```

```python
for column in numeric_columns:
    
    plt.figure(figsize=(8, 5))
    
    sns.kdeplot(
        data=df,
        x=column,
        fill=True
    )
    
    plt.title(f"Density Plot of {column}")
    plt.xlabel(column)
    plt.ylabel("Density")
    
    plt.show()
```

## Box Plot

## Categorical Variables

Analyze:

### Frequency
### Unique values
### Value counts
### Percentage distribution

```python
categorical_columns = [
    "Property_Type",
    "Furnishing_Status",
    "Construction_Status",
    "Locality_Type"
]
```

### 1. Frequency
Own Words; Frequency means the number of times each category appears in a column.

Example; For Property_Type:

Apartment          18
Independent House  17
Villa              10


```python
categorical_columns = [
    "Property_Type",
    "Furnishing_Status",
    "Construction_Status",
    "Locality_Type"
]
```

```python

for column in categorical_columns:
    print("\n", column)

    print(df[column].value_counts())
```

```python
df["Property_Type"].value_counts()
```

```python
### 2. Unique Values
Own Words;  Unique values tell us what different categories are present in a column.
```

```python
for colum in categorical_columns:
    print("\n", colum)
    print(df[colum].nunique())
```

```python
df["Construction_Status"].nunique()
```

```python
### 4. Value Counts

value_counts() gives the frequency of each category.

```

```python
for column in categorical_columns:
    print(f"\nValue Counts - {column}")
    print(df[column].value_counts())
```

```python
df["Furnishing_Status"].value_counts()
```

```python
### Percentage Distribution
```

```python
for column in categorical_columns:
    
    percentage = (
        df[column]
        .value_counts(normalize=True)
        .mul(100)
        .round(2)
    )
    
    print(f"\nPercentage Distribution - {column}")
    print(percentage)
```

```python
df["Property_Type"].value_counts(normalize=True).mul(100).round(2)
```

### Visualizations:
### Bar Chart
### Count Plot
### Pie Chart where appropriate


```python
1. Bar Chart
Step 1 – Understand the Concept
Own Words; A Bar Chart displays the frequency of each category using bars.
```

```python
property_counts = df["Property_Type"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    property_counts.index,
    property_counts.values
)

plt.title("Property Type Distribution")
plt.xlabel("Property Type")
plt.ylabel("Frequency")

plt.xticks(rotation=30)

plt.show()
```

```python

plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="Property_Type"
)

plt.title("Property Type Count")
plt.xlabel("Property Type")
plt.ylabel("Count")

plt.xticks(rotation=30)

plt.show()
```

```python
Pie Chart
Step 1 – Understand the Concept

A Pie Chart shows how categories contribute to the total as proportions or percentages.
```

```python
property_counts = df["Property_Type"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    property_counts.values,
    labels=property_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Property Type Percentage Distribution")

plt.show()
```

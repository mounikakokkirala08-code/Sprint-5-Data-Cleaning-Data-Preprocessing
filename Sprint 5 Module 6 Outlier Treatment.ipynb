# Outlier Treatment

## What is an Outlier?

### Step 1 – Understand the Concept (Markdown)
Definition: An outlier is a data value that is significantly different from the other values in a dataset.

In simple words: Outlier = An unusually high or low value.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Age: 20, 22, 25, 24, 23, 150
                         ↑
                      Outlier
### Step 3 – Implement the Concept (Python)


```python
import pandas as pd


df = pd.DataFrame({
    "Age": [20, 22, 25, 24, 23, 150]
})


# Check basic statistics
print(df["Age"].describe())
```

### Step 1 – Understand the Concept (Markdown)

Outlier = Unusual value

## IQR Method
### Step 1 – Understand the Concept (Markdown)
Definition:The IQR (Interquartile Range) Method is a statistical technique used to detect outliers.

In simple words: IQR Method = Find values that are unusually far from the middle 50% of the data.

Formula:
IQR = Q3 - Q1


Lower Bound = Q1 - 1.5 × IQR
Upper Bound = Q3 + 1.5 × IQR

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Data: 10, 12, 15, 18, 20, 22, 50
                         ↑
                      Outlier
### Step 3 – Implement the Concept (Python)

```python
Q1 = df["Age"].quantile(0.25)
Q3 = df["Age"].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

# Find outliers
outliers = df[
    (df["Age"] < lower) |
    (df["Age"] > upper)
]

print(outliers)
```

## Z-Score Method
### Step 1 – Understand the Concept (Markdown)

In simple words: Z-Score = How far a value is from the average
### Step 2 – Demonstrate the Concept (Markdown)

Data: 20, 22, 25, 24, 23, 100

If 100 has a Z-score greater than 3, it is considered an outlier.

## Percentile Method
### Step 1 – Understand the Concept (Markdown)
Definition:The Percentile Method detects outliers by identifying values that fall below a lower percentile or above an upper percentile.

In simple words: Percentile Method = Remove extreme values from the top and bottom percentages.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Suppose we use:

Lower limit = 5th percentile
Upper limit = 95th percentile
5%          90%           5%
|------------|-------------|
Outliers      Normal       Outliers

Values below the 5th percentile or above the 95th percentile are treated as potential outliers.
### Step 3 – Implement the Concept (Python)


```python
import pandas as pd


df = pd.DataFrame({
    "Age": [20, 22, 25, 30, 35, 40, 100]
})


# Find percentile limits
lower = df["Age"].quantile(0.05)
upper = df["Age"].quantile(0.95)


# Find outliers
outliers = df[
    (df["Age"] < lower) |
    (df["Age"] > upper)
]


print(outliers)
```

## Winsorization
### Step 1 – Understand the Concept (Markdown)
Definition: Winsorization is an outlier treatment technique that replaces extreme values with specified percentile values instead of removing them.

In simple words: Winsorization = Cap extreme values instead of deleting them.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example: Suppose we use 5th and 95th percentiles:

Before:
10, 20, 25, 30, 35, 100


After Winsorization:
15, 20, 25, 30, 35, 90

### Step 3 – Implement the Concept (Python)


```python
import pandas as pd


df = pd.DataFrame({
    "Age": [10, 20, 25, 30, 35, 100]
})


lower = df["Age"].quantile(0.05)
upper = df["Age"].quantile(0.95)


# Winsorization
df["Age"] = df["Age"].clip(lower, upper)


print(df)
```

## Capping
### Step 1 – Understand the Concept (Markdown)
Definition: Capping is an outlier treatment technique where extreme values are replaced with a predefined upper or lower limit.

In simple words:Capping = Limit extreme values instead of removing them.
### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Suppose the allowed range is 10 to 100:

Before: 5, 20, 40, 80, 150


After:  10, 20, 40, 80, 100
        ↑              ↑
      Lower          Upper
      cap            cap
5 → replaced with 10
150 → replaced with 100

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [5, 20, 40, 80, 150]
})

# Capping
df["Age"] = df["Age"].clip(lower=10, upper=100)

print(df)
```

## Flooring
### Step 1 – Understand the Concept (Markdown)
Definition: Flooring is an outlier treatment technique where values below a specified lower limit are replaced with that minimum limit.

In simple words: Flooring = Set a minimum limit for low values.
### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Suppose the lower limit is 10:

Before: 5, 8, 15, 20, 30


After:  10, 10, 15, 20, 30
         ↑   ↑
       Floored values
### Step 3 – Implement the Concept (Python)


```python
import pandas as pd


df = pd.DataFrame({
    "Age": [5, 8, 15, 20, 30]
})


# Flooring
df["Age"] = df["Age"].clip(lower=10)


print(df)
```

Definition: Retaining outliers means keeping extreme values because they may represent real and important information.

In simple words: Retaining Outliers = Keep unusual values when they are meaningful.

### Step 2 – Demonstrate the Concept (Markdown)

A salary of 500K may be a genuine high salary, so removing it could lose useful information.

# 10 – Target Variable Analysis

### Step 1 – Understand the Concept (Markdown)


```python
Target Variable

The target variable is the variable that we want our Machine Learning model to predict.

In this house-price dataset, the target variable is:

Price_lakhs

Because Price_lakhs contains continuous numerical values, this is a regression problem.

Why is Target Analysis Important?

Before building a Machine Learning model, we need to understand the target variable because:

It tells us what the model needs to predict.
It helps us determine whether the problem is classification or regression.
It shows the typical target value using the mean and median.
It tells us how spread out the target values are using variance and standard deviation.
Skewness tells us whether the target distribution is symmetric or heavily concentrated on one side.
Detecting outliers is important because extreme target values can strongly affect regression models.
Understanding the target distribution helps us decide whether transformations such as log1p() may be useful.
It helps us understand whether the dataset contains enough variation for meaningful prediction.
Classification vs Regression

Classification:

The target contains categories/classes.

Example:

Purchased
Yes
No
Yes
No

For classification, we analyze:

Class Distribution
Class Counts
Class Percentages
Class Imbalance

Regression:

The target contains continuous numerical values.

Example:

Price_lakhs
45.5
52.0
75.5
120.0

For regression, we analyze:

Target Distribution
Mean
Median
Variance
Standard Deviation
Skewness
Outliers
```

## Step 2 – Demonstrate the Concept (Markdown)


```python
Simple Example

Suppose the house prices are:

40, 45, 50, 55, 60, 65, 70, 150

The mean will be higher because 150 is much larger than the other values.

The median represents the middle value and is less affected by the extreme value.

The value 150 may be an outlier.

Therefore, before training a regression model, we should understand:

What is the typical price?
How much do prices vary?
Is the distribution skewed?
Are there extreme prices?

For example:

Mean              → Average house price
Median            → Middle house price
Variance          → Spread of prices
Standard Deviation → Typical amount of variation
Skewness          → Shape/asymmetry of distribution
Outliers          → Unusually high or low prices
```

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("house_price_prediction_cleaned.csv")

# Display first 5 rows
df.head()
```

## 2. Identify the Target Variable

```python
target = "Price_lakhs"

print("Target Variable:", target)
print("Data Type:", df[target].dtype)
```

###  Mean

```python
mean_value = df[target].mean()

print("Mean:", mean_value)
```

### Median

```python
median_value = df[target].median()

print("Median:", median_value)
```

## Variance

```python
Variance = df[target].var()
print(Variance)
```

## Standard Deviation

```python
STD = df[target].std()
print(STD)
```

## Skewness

```python
skewness_value = df[target].skew()

print("Skewness:", skewness_value)
```

## Outliers

```python
Q3 = np.percentile(df[target], 75)
Q1 = np.percentile(df[target], 25)

IQR = Q3-Q1
print(IQR)

Lower_limit = Q1 - 1.5 * IQR
Higher_limit = Q3 + 1.5 * IQR


outlier = df[ (df[target] < Lower_limit) | (df[target] > Higher_limit)]
print(outlier)

```

```python
print("Number of Outliers:", len(outlier))

display(outlier[[target]])
```

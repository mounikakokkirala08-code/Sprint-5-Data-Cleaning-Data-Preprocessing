# Matplotlib & Seaborn


### Step 1 – Understand the Concept (Markdown)

```python
# Matplotlib

Matplotlib is a Python library used to create data visualizations. It helps us understand data using different types of charts and graphs.

Matplotlib can be used to create:

Line plots, Bar charts, Histograms, Scatter plots, Box plots, Pie charts

It provides detailed control over the chart, including titles, labels, axes, figure size, and layout.

# Seaborn

Seaborn is a Python visualization library built on top of Matplotlib. It is mainly used for creating statistical and attractive visualizations.

Seaborn can be used to create:

Histograms
Count plots
Box plots
Scatter plots
Bar plots
Heatmaps
Violin plots
Pair plots

Seaborn works very well with Pandas DataFrames.

Why Visualization is Important Before Machine Learning

Data visualization helps us:

Understand the distribution of data.
    
Identify patterns and relationships.
Detect outliers.
Identify skewed distributions.
Understand categorical variables.
Compare numerical variables.
Find possible relationships between features and the target variable.
```

## Step 2 – Demonstrate the Concept (Markdown)


```python
# Simple Example – Matplotlib

# Suppose we have house prices:

# Price = 40, 50, 60, 70, 80

# We can visualize them using a line plot.

import matplotlib.pyplot as plt


price = [40, 50, 60, 70, 80]


plt.plot(price)


plt.title("House Price")
plt.xlabel("House")
plt.ylabel("Price in Lakhs")


plt.show()
```

## Step 3 – Implement the Concept (Python)


```python
import pandas as pd
import numpy as np


import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# 2. Load Dataset
df = pd.read_csv("house_price_prediction_cleaned.csv")


df.head()
```

```python
#3. Matplotlib – Line Plot
plt.figure(figsize=(8, 5))


plt.plot(df["Price_lakhs"])


plt.title("House Price Line Plot")
plt.xlabel("House Index")
plt.ylabel("Price in Lakhs")

```

```python
#4. Matplotlib – Histogram
plt.figure(figsize=(8, 5))


plt.hist(df["Price_lakhs"], bins=10)


plt.title("House Price Distribution")
plt.xlabel("Price in Lakhs")
plt.ylabel("Frequency")


plt.show()
```

```python
# Matplotlib – Scatter Plot
plt.figure(figsize=(8, 5))


plt.scatter(
    df["Area_sqft"],
    df["Price_lakhs"]
)


plt.title("Area vs House Price")
plt.xlabel("Area in Sqft")
plt.ylabel("Price in Lakhs")


plt.show()
```

```python

## 6. Seaborn – Histogram
plt.figure(figsize=(8, 5))


sns.histplot(
    data=df,
    x="Price_lakhs",
    kde=True
)


plt.title("House Price Distribution")
plt.xlabel("Price in Lakhs")
plt.ylabel("Frequency")


plt.show()
```

```python
#7. Seaborn – Count Plot
plt.figure(figsize=(8, 5))


sns.countplot(
    data=df,
    x="Property_Type"
)


plt.title("Property Type Distribution")
plt.xlabel("Property Type")
plt.ylabel("Count")


plt.xticks(rotation=30)


plt.show()
```

```python
# 8. Seaborn – Box Plot
plt.figure(figsize=(8, 5))


sns.boxplot(
    data=df,
    x="Property_Type",
    y="Price_lakhs"
)


plt.title("Property Type vs House Price")
plt.xlabel("Property Type")
plt.ylabel("Price in Lakhs")


plt.xticks(rotation=30)


plt.show()
```

```python
# 9. Seaborn – Scatter Plot
plt.figure(figsize=(8, 5))


sns.scatterplot(
    data=df,
    x="Area_sqft",
    y="Price_lakhs"
)


plt.title("Area vs House Price")
plt.xlabel("Area in Sqft")
plt.ylabel("Price in Lakhs")


plt.show()
```

```python
#10. Seaborn – Heatmap
plt.figure(figsize=(10, 6))

correlation = df.select_dtypes(
    include="number"
).corr()


sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")


plt.show()
```

```python
# 11. Seaborn – Pair Plot
sns.pairplot(
    df.select_dtypes(include="number")
)


plt.show()
```

```python


# Final Conclusion

# Matplotlib provides detailed control over visualizations, while Seaborn makes statistical visualizations easier and more informative.

# For this house-price project:

# Histogram → Understand price distribution
# Count Plot → Analyze categorical variables
# Box Plot → Detect outliers and compare categories
# Scatter Plot → Analyze relationships between numerical features
# Heatmap → Understand correlations between numerical variables
# Pair Plot → Analyze relationships among multiple numerical variables
```

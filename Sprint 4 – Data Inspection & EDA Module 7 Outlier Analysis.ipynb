# Outlier Analysis

```python
Step 1 – Understand the Concept
1. What is an Outlier?
Own Words

An outlier is an observation that is unusually far away from most other observations.

For example, if most houses have:

Price = 50–200 lakhs

but one genuine house has:

Price = 500 lakhs

that observation may be an outlier.

Definition

An outlier is a data point that differs significantly from the general pattern or distribution of the dataset.

2. Why Do Outliers Occur?

Outliers can occur because of:

1. Data Entry Error

Example:

Area_sqft = 25000

when the actual value should have been:

Area_sqft = 2500
2. Measurement Error

A sensor or measurement system may record an incorrect value.

3. Genuine Extreme Observation

The observation can be completely correct.

Example:

Area_sqft = 10000
Price_lakhs = 900

This could be a genuine luxury property.

4. Rare Events

Some houses may have unusual characteristics because they are:

Luxury property
Prime location
Large villa
Special construction
5. Data Processing Problems

Outliers can also appear because of:

Incorrect unit
Incorrect conversion
Incorrect data type
Incorrect merging
3. Types of Outliers

There are several types, but for your notebook focus on:

1. Univariate Outliers
2. Multivariate Outliers
```

```python
Univariate Outliers
Step 1 – Understand the Concept
Own Words

A Univariate Outlier is an unusual value detected by looking at one variable at a time.

Example:

Price_lakhs

If most values are between:

40 – 200

but one value is:

350

it may be a univariate outlier.

Step 2 – Real-World Analogy

Imagine the house prices are:

50
55
60
65
70
75
80
85
90
400

The 400 value is much higher than the other values.

Step 3 – Implement

We'll use:

IQR
Z-Score
Box Plot
```

```python
5. Multivariate Outliers
Step 1 – Understand the Concept
Own Words

A Multivariate Outlier may not look unusual when looking at one variable individually, but becomes unusual when multiple variables are considered together.

Step 2 – Real-World Analogy

Suppose:

Area_sqft = 3000
Price_lakhs = 50

Individually:

3000 sqft → possible
50 lakhs  → possible

But together:

3000 sqft + ₹50 lakhs

may be unusual compared with the overall relationship between area and price.

Step 3 – Implement

Use a scatter plot:

sns.scatterplot(
    data=df,
    x="Area_sqft",
    y="Price_lakhs"
)

plt.title("Area vs Price - Multivariate Outlier Detection")
plt.show()

Points far away from the main pattern should be investigated.
```

```python
6. IQR Method
Step 1 – Understand the Concept
Own Words

IQR (Interquartile Range) measures the spread of the middle 50% of the data.

Formula
IQR = Q3 - Q1

Where:

Q1 = 25th percentile
Q3 = 75th percentile

Outlier boundaries:

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these boundaries are potential outliers.

Step 2 – Real-World Analogy

Suppose:

Q1 = 70
Q3 = 150

Then:

IQR = 150 - 70
    = 80

Lower boundary:

70 - (1.5 × 80)
= -50

Upper boundary:

150 + (1.5 × 80)
= 270

So values above 270 are potential outliers.

Step 3 – Implement
IQR for Price
Q1 = df["Price_lakhs"].quantile(0.25)
Q3 = df["Price_lakhs"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
8. Z-Score Method
Step 1 – Understand the Concept
Own Words

The Z-score tells us how far a value is from the mean in terms of standard deviations.

Formula
Z = (X - Mean) / Standard Deviation

Common rule:

|Z| > 3

→ Potential outlier.

Step 2 – Real-World Analogy

Suppose average house price is:

Mean = ₹100 lakhs

and standard deviation is:

Std = ₹20 lakhs

A house costing:

₹180 lakhs

has:

Z = (180 - 100) / 20
  = 4

So it is a potential outlier using the |Z| > 3 rule.

Step 3 – Implement
from scipy.stats import zscore

df["Price_Zscore"] = zscore(
    df["Price_lakhs"]
)

df[
    df["Price_Zscore"].abs() > 3
]
```

```python
9. Z-Score for Multiple Features
zscore_outliers = {}

for column in numeric_columns:

    z_scores = zscore(
        df[column].dropna()
    )

    zscore_outliers[column] = (
        abs(z_scores) > 3
    ).sum()

zscore_outliers

A cleaner summary:

zscore_summary = []

for column in numeric_columns:

    values = df[column].dropna()

    z_scores = zscore(values)

    count = (abs(z_scores) > 3).sum()

    zscore_summary.append({
        "Column": column,
        "ZScore_Outlier_Count": count
    })

zscore_summary = pd.DataFrame(
    zscore_summary
)

zscore_summary
```

```python
10. Box Plot
Step 1 – Understand the Concept
Own Words

A Box Plot is a visual method for identifying potential outliers.

Potential outliers are often shown as points beyond the whiskers.

Step 2 – Real-World Analogy

If most house prices are concentrated in one region but a few points are far away, those points need investigation.

Step 3 – Implement
plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    y="Price_lakhs"
)

plt.title("Box Plot of House Prices")
plt.ylabel("Price (Lakhs)")

plt.show()
Box Plots for All Numerical Features
for column in numeric_columns:

    plt.figure(figsize=(7, 4))

    sns.boxplot(
        data=df,
        y=column
    )

    plt.title(f"Box Plot - {column}")

    plt.show()
```

## Feature Selection
### Step 1 – Understand the Concept (Markdown)

Definition:Feature Selection is the process of selecting the most important features from a dataset and removing irrelevant or unnecessary features.

In simple words:Instead of giving all columns to the ML model, we keep only the useful columns.

###Step 2 – Demonstrate the Concept (Markdown)

Example: House Price Prediction
Features:
Area_sqft        → Important ✅
Bedrooms         → Important ✅
Location         → Important ✅
House_ID         → Not useful ❌
Owner_Name       → Not useful ❌

So we select:

Area_sqft, Bedrooms, Location

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd
from sklearn.feature_selection import SelectKBest, f_regression


X = pd.DataFrame({
    "Area": [1000, 1500, 2000, 2500, 3000],
    "Bedrooms": [2, 3, 3, 4, 5],
    "Age": [10, 8, 5, 3, 1],
    "Noise": [100, 50, 200, 20, 300]
})

y = [30, 45, 60, 80, 100]


selecter = SelectKBest(score_func = f_regression, k=1)

x_selected = selecter.fit_transform(X,y)

x_selected
```

## Relevant Features
### Step 1 – Understand the Concept (Markdown)

Definition: Relevant features are the features that have useful information or a meaningful relationship with the target variable.

In simple words: They are the columns that help the ML model make better predictions.

Relevant features can improve model performance and reduce unnecessary information.

## Redundant Features
### Step 1 – Understand the Concept (Markdown)

Definition: Redundant features are features that provide duplicate or very similar information.

In simple words:Two or more columns tell the model almost the same thing.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Area_sqft        → 2000
Area_sqft_m2     → 185.8

Both represent the same property area, so one may be redundant.

Another example:

Age_years
Construction_Year

These can contain almost the same information.

Problems:

Increases model complexity
Can cause multicollinearity
May reduce model interpretability
Adds unnecessary information

## Univariate Feature Selection
### Step 1 – Understand the Concept (Markdown)

Definition: Univariate Feature Selection evaluates each feature individually against the target and selects the most relevant features.

In simple words:
It checks one feature at a time to see how strongly it is related to the target.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Feature 1 → Target → Strong relationship ✅
Feature 2 → Target → Weak relationship ❌
Feature 3 → Target → Strong relationship ✅

Common methods:

f_classif → Classification
f_regression → Regression
chi2 → Categorical features
mutual_info_classif → Classification

### Step 3 – Implement the Concept (Python)

## Mutual Information
### Step 1 – Understand the Concept (Markdown)

Definition: Mutual Information (MI) measures how much information one feature provides about the target variable.

In simple words: It tells us whether a feature is useful for predicting the target, including relationships that may not be linear.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Feature        MI Score


Area_sqft      0.80  → Very useful ✅
Bedrooms       0.55  → Useful ✅
Age            0.10  → Less useful
House_ID       0.00  → Not useful ❌

Important:

Higher MI score → More information about the target → More useful feature.

It can detect non-linear relationships, unlike simple correlation.

## Recursive Feature Elimination (RFE)
### Step 1 – Understand the Concept (Markdown)

Definition; RFE is a feature selection technique that recursively removes the least important features until the required number of features remains.

In simple words: It starts with all features → removes the weakest feature → repeats until only the best features remain.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

10 Features
     ↓
Remove least important
     ↓
9 Features
     ↓
Remove least important
     ↓
...
     ↓
5 Best Features

RFE needs a model/estimator to determine feature importance

### Step 3 – Implement the Concept (Python)



```python
from sklearn.feature_selection import RFE
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

selector = RFE(
    estimator=model,
    n_features_to_select=3
)

X_selected = selector.fit_transform(X, y)

print(X_selected)
```

## Feature Importance
### Step 1 – Understand the Concept (Markdown)

Definition: Feature Importance tells us how important each feature is for making predictions in a Machine Learning model.

In simple words: It tells us which columns are most useful to the model.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Area_sqft       → 0.45 → Important ✅
Location_Score  → 0.30 → Important ✅
Bedrooms        → 0.15 → Less important
Age_years       → 0.10 → Less important

Higher score → More important feature.

Common models that provide feature importance:

Decision Tree
Random Forest
XGBoost
LightGBM
### Step 3 – Implement the Concept (Python)

```python
from sklearn.ensemble import RandomForestRegressor

model = RandomForestRegressor(random_state=42)

model.fit(X, y)

importance = model.feature_importances_

print(importance)
```

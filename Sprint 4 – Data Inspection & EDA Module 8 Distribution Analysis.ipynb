```python
Normal Distribution
Own Words

A Normal Distribution is a bell-shaped distribution where most observations are around the center, and fewer observations occur toward the extremes.

μ
μ−1σ
μ+1σ
−1≤Z≤1
About 68.3% of values fall in this interval, from 1 below the mean to 1 above it
z
z
Learn more
Characteristics
Mean ≈ Median ≈ Mode
        ↓
Bell-shaped curve
        ↓
Symmetric
Example

A feature might approximately follow:

        /\
       /  \
      /    \
_____/      \_____
Python Example
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(
    loc=100,
    scale=15,
    size=1000
)

plt.figure(figsize=(8, 5))

plt.hist(
    data,
    bins=30,
    edgecolor="black"
)

plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
Uniform Distribution
Own Words

In a Uniform Distribution, values have approximately equal probability across a specified range.

Example
10 ───────────────────── 20

Values between 10 and 20 are approximately equally likely.

Shape
 ____________
|            |
|            |
|____________|
Python Example
data = np.random.uniform(
    low=10,
    high=20,
    size=1000
)

plt.figure(figsize=(8, 5))

plt.hist(
    data,
    bins=20,
    edgecolor="black"
)

plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.show()
```

```python
Kurtosis
Own Words

Kurtosis describes the shape of the distribution, particularly the degree of tail heaviness compared with a normal distribution.

For practical data analysis, pandas .kurt() reports excess kurtosis, where approximately:

Kurtosis ≈ 0
→ Similar tail behavior to normal distribution

Kurtosis > 0
→ Heavier tails / more extreme observations

Kurtosis < 0
→ Lighter tails / fewer extreme observations
Python
df["Price_lakhs"].kurt()
```

## Square Root Transformation
### Step 1 – Understand the Concept (Markdown)

Definition:Square Root Transformation converts numerical values using the square root function to reduce skewness and the influence of large values.

In simple words:
It reduces the difference between small and large values and can make right-skewed data more balanced.

### Step 2 – Demonstrate the Concept (Markdown)

Simple example:

Original:     4    16    100
Square Root:  2     4     10

The large value 100 is reduced to 10.

Useful for:

Right-skewed data
Count data
Moderate outliers
Variables such as number of customers, visits, or transactions

```python
import numpy as np

data = [4, 16, 100]

transformed_data = np.sqrt(data)

print(transformed_data)
```

## Power Transformation
### Step 1 – Understand the Concept (Markdown)

Definition: Power Transformation changes numerical data using a power function to make the data more normally distributed and reduce skewness.

In simple words:It changes the shape of the data so that it becomes more balanced and suitable for ML models.

### Step 2 – Demonstrate the Concept (Markdown)

Common power transformations:

Box-Cox → works with positive values only
Yeo-Johnson → works with positive, zero, and negative values

Example:

Original data → Highly skewed
                 ↓
          Power Transformation
                 ↓
        More balanced distribution

Useful for:

Reducing skewness
Making data more normally distributed
Improving some ML models

```python
from sklearn.preprocessing import PowerTransformer

data = [[10], [20], [50], [100], [1000]]

transformer = PowerTransformer(method="yeo-johnson")

result = transformer.fit_transform(data)

print(result)
```

## 1Box-Cox Transformation
### Step 1 – Understand the Concept (Markdown)

Definition: Box-Cox Transformation is a power transformation used to reduce skewness and make numerical data closer to a normal distribution.

In simple words: It changes the shape of the data to make it more balanced.

Important:
Box-Cox works only with positive values (> 0).

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Original:
10, 20, 50, 100, 1000
       ↓
   Box-Cox
       ↓
More balanced distribution

Main uses:

Reduce skewness
Make data more normally distributed
Improve some statistical/ML models
### Step 3 – Implement the Concept (Python)


Remember:

Box-Cox → Power Transformation → Positive values only → Reduces skewness.

Difference:

Box-Cox      → Positive values only
Yeo-Johnson  → Positive + Zero + Negative values

```python
from sklearn.preprocessing import PowerTransformer


data = [[10], [20], [50], [100], [1000]]


transformer = PowerTransformer(method="box-cox")


result = transformer.fit_transform(data)


print(result)
```

## 2Yeo-Johnson Transformation
### Step 1 – Understand the Concept (Markdown)

Definition: Yeo-Johnson Transformation is a power transformation used to reduce skewness and make data more normally distributed.

In simple words:It is similar to Box-Cox, but it can handle positive, zero, and negative values.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Original:
-100, -10, 0, 10, 100
          ↓
    Yeo-Johnson
          ↓
More balanced distribution

Main advantage:

    Main advantage:

Box-Cox      → Positive values only
Yeo-Johnson  → Positive + Zero + Negative
### Step 3 – Implement the Concept (Python)

       Remember:

Yeo-Johnson → Power Transformation → Handles negative, zero, and positive values → Reduces skewness. 

```python
from sklearn.preprocessing import PowerTransformer

data = [[-100], [-10], [0], [10], [100]]

transformer = PowerTransformer(method="yeo-johnson")

result = transformer.fit_transform(data)

print(result)
```

## Handling Skewed Data
### Step 1 – Understand the Concept (Markdown)

Definition:Handling skewed data means using techniques to reduce the imbalance in a data distribution so it becomes more suitable for analysis and ML.

In simple words: If most values are on one side and a few extreme values are on the other side, the data is skewed.

### Step 2 – Demonstrate the Concept (Markdown)

Types: 
Right Skewed (Positive Skew) → Long tail on the right
Left Skewed (Negative Skew) → Long tail on the left

Common techniques:

Technique	Best Used For
Log Transformation	Right-skewed data
Square Root	Moderate right skew
Box-Cox	Positive data
Yeo-Johnson	Positive, zero & negative data

### Step 3 – Implement the Concept (Python)

```python
import numpy as np

data = [10, 20, 50, 100, 1000]

# Log transformation
transformed = np.log1p(data)

print(transformed)
```

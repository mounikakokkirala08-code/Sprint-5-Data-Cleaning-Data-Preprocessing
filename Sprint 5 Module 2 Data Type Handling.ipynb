# Data Type Handling

## Identifying Data Types
### Step 1 – Understand the Concept (Markdown)
Definition: Identifying data types means determining whether each column contains numerical, categorical, text, date, or other types of data.

In simple words: Before preprocessing, we need to know what type of data each column contains.

Common data types:

Numerical → Age, Salary, Height
Categorical → Gender, City, Department
Text → Name, Comments, Reviews
Date/Time → Birth Date, Joining Date
Boolean → True/False

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Column	Data Type
Age	Numerical
Salary	Numerical
Gender	Categorical

### Step 3 – Implement the Concept (Python)


```python
import pandas as pd

data = {
    "name" : ["mani", "roja", "mahi"],
    "age" : [20,30,40],
    "salary" : [30000,40000,30000]
}
df = pd.DataFrame(data)


# Check data types
print(df.dtypes)


# Detailed information
print(df.info())
```

## Converting Data Types
### Step 1 – Understand the Concept (Markdown)
Definition:Data type conversion means changing data from one data type to another so it can be used correctly for analysis or machine learning.

In simple words:Data Type Conversion = Changing the format/type of data.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
"25" → 25          String → Integer
"30000" → 30000    String → Integer
"2026-08-19" → Date Date String → DateTime
### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

data = {
    "age": ["20", "30", "40"],
    "salary": ["30000", "40000", "50000"],
    "joining_date": ["2026-01-10", "2026-02-15", "2026-03-20"]
}

df = pd.DataFrame(data)
print(df.dtypes)
# String → Integer
df["age"] = df["age"].astype(int)
df["salary"] = df["salary"].astype(int)

df["joining_date"] = pd.to_datetime(df["joining_date"])


print(df.dtypes)
```

## astype()
### Step 1 – Understand the Concept (Markdown)
Definition ; astype() is a Pandas method used to change the data type of a column or DataFrame.

In simple words:

astype() = Convert data from one type to another.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Suppose Age is stored as text:

"20"
"30"
"40"

We can convert it to integers:

astype() → Used to change the data type of Pandas columns.

```python
import pandas as pd

df = pd.DataFrame({
    "age": ["20", "30", "40"],
    "salary": ["30000", "40000", "50000"]
})

# Convert string to integer
df["age"] = df["age"].astype(int)

# Convert string to float
df["salary"] = df["salary"].astype(float)

print(df)
print(df.dtypes)
```

## to_numeric()
### Step 1 – Understand the Concept (Markdown)
Definition: pd.to_numeric() is a Pandas function used to convert values into numeric data types such as integers or floats.

In simple words: to_numeric() = Convert values into numbers.

It is especially useful when data contains invalid or non-numeric values.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
"100"  → 100
"250"  → 250
"abc"  → NaN

Using errors="coerce" converts invalid values into NaN instead of producing an error.

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "salary": ["30000", "40000", "abc", "50000"]
})

# Convert to numeric
df["salary"] = pd.to_numeric(
    df["salary"],     errors="coerce"
)

print(df)
print(df.dtypes)
# pd.to_numeric() → Converts data into numeric values and can safely handle invalid values.
```

## to_datetime()
### Step 1 – Understand the Concept (Markdown)
Definition:pd.to_datetime() is a Pandas function used to convert values into Date/Time format.

In simple words:to_datetime() = Convert text or other values into date/time data.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
"2026-01-10" → 2026-01-10
"15-02-2026" → 2026-02-15

Once converted, we can easily extract year, month, day, day of week, etc.

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "joining_date": ["2026-01-10", "2026-02-15", "2026-03-20"]
})

# Convert to DateTime
df["joining_date"] = pd.to_datetime(df["joining_date"])

print(df)
print(df.dtypes)
```

```python
df["year"] = df["joining_date"].dt.year
df["month"] = df["joining_date"].dt.month
df["day"] = df["joining_date"].dt.day

print(df)
```

## Handling Invalid Conversions
### Step 1 – Understand the Concept (Markdown)
Definition: Handling invalid conversions means managing values that cannot be converted into the required data type without causing errors.

In simple words; Invalid Conversion = A value cannot be converted → Handle it safely.

Common methods:

errors="coerce" → Invalid values become NaN
errors="ignore" → Keep the original value
errors="raise" → Show an error
### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Suppose we have:

Age
20
30
abc
40

"abc" cannot be converted to a number.

Using:

errors="coerce"

we get:

20
30
NaN
40

Then we can handle the NaN value separately.

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "age": ["20", "30", "abc", "40"]
})

# Convert invalid values to NaN
df["age"] = pd.to_numeric(
    df["age"],
    errors="coerce"
)

# Handle NaN
df["age"] = df["age"].fillna(df["age"].median())

print(df)
```

## Detecting Incorrect Data Types
### Step 1 – Understand the Concept (Markdown)
Definition:Detecting incorrect data types means identifying columns whose data type does not match the actual data they contain.

In simple words: Check whether each column has the correct data type before analysis.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example: Suppose age contains:

"20"
"30"
"40"

Although these look like numbers, Pandas may store them as string/object.

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "age": ["20", "30", "40"],
    "salary": [30000, 40000, 50000]
})

# Check data types
print(df.dtypes)

# Convert incorrect data type
df["age"] = pd.to_numeric(df["age"])

print("\nAfter conversion:")
print(df.dtypes)
```

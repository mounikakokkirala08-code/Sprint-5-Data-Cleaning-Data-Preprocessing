# Encoding Categorical Variables

## What is Categorical Encoding?
### Step 1 – Understand the Concept (Markdown)
Definition: Categorical Encoding is the process of converting categorical/text data into numerical values so that machine learning algorithms can understand it.

In simple words: Categorical Encoding = Convert categories into numbers.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Before:

Gender
Male
Female
Male

After encoding:

Gender
1
0

Here:

Male   → 1
Female → 0
Common Encoding Techniques:
Label Encoding
One-Hot Encoding
Ordinal Encoding
Target Encoding

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "name": ["raju", "mahi"],
    "Gender": ["Male", "Female"],
    "Age" : [20, 30]})
```

```python
from sklearn.preprocessing import LabelEncoder

df["Gender"] = LabelEncoder().fit_transform(df["Gender"])

df
```

## Types of Categorical Encoding
### Step 1 – Understand the Concept (Markdown)
Definition: Categorical encoding converts categorical values into numerical values for machine learning.

Main Types:
Label Encoding
One-Hot Encoding
Ordinal Encoding
Binary Encoding
Frequency Encoding
Target Encoding
### Step 2 – Demonstrate the Concept (Markdown)
xample:
Gender
Male

Male

Different encoding techniques convert this data differently:

Label Encoding    → Male = 1, Female = 0
One-Hot Encoding  → Male = [1,0], Female = [0,1]
Ordinal Encoding  → Low = 1, Medium = 2, High = 3  
### Step 3 – Implement the Concept (Python)

```python
Label Encoding     → Categories → Numbers
One-Hot Encoding   → Categories → 0/1 columns
Ordinal Encoding   → Ordered categories → Numbers
Binary Encoding    → Categories → Binary values
Frequency Encoding → Categories → Frequency
Target Encoding    → Categories → Target statistics
```

```python
## Nominal Variables
### Step 1 – Understand the Concept (Markdown)
Definition: Nominal variables are categorical variables that contain different categories with no natural order or ranking.

In simple words: Nominal = Categories with NO order.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Color
Red
Blue
Green
### Step 3 – Implement the Concept (Python)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red"]
})

print(df["Color"])

#one hot encoding

df = pd.get_dummies(df, columns=["Color"])

print(df)
```

## Ordinal Variables
### Step 1 – Understand the Concept (Markdown)
Definition: Ordinal variables are categorical variables that have a natural order or ranking, but the difference between categories is not necessarily equal.

In simple words:Ordinal = Categories with an order.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Education Level

School < Graduate < Postgraduate

There is a clear order:

Low < Medium < High

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Level": ["Low", "Medium", "High", "Low"]
})

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

df["Level_Encoded"] = encoder.fit_transform(
    df[["Level"]]
)

print(df)
```

```python
## Binary Variables
### Step 1 – Understand the Concept (Markdown)
Definition:Binary variables are variables that have only two possible values or categories.

In simple words: Binary = Only two possible values.

### Step 2 – Demonstrate the Concept (Markdown)
Simple examples:
Gender     → Male / Female
Status     → Yes / No
Pass       → True / False
Payment    → Paid / Unpaid

### Step 3 – Implement the Concept (Python)
```

```python
import pandas as pd

df = pd.DataFrame({
    "Status": ["Yes", "No", "Yes", "No"]
})

# Binary encoding
df["Status"] = df["Status"].map({
    "No": 0,
    "Yes": 1
})

print(df)
```

## Ordinal Encoding
### Step 1 – Understand the Concept (Markdown)
Definition:Ordinal Encoding converts categorical values into numbers while preserving their natural order or ranking.

In simple words: Ordinal Encoding = Ordered Categories → Ordered Numbers

Step 2 – Demonstrate the Concept (Markdown)
Example:
Level
Low < Medium < High

After encoding:

Low    → 0
### Step 3 – Implement the Concept (Python)

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.DataFrame({
    "Level": ["Low", "Medium", "High", "Low"]
})

encoder = OrdinalEncoder(
    categories=[["Low", "Medium", "High"]]
)

df["Level_Encoded"] = encoder.fit_transform(
    df[["Level"]]
)

print(df)
```

## Dummy Variables
### Step 1 – Understand the Concept (Markdown)
Definition: Dummy variables are binary (0/1) variables created from categorical data for machine learning.

In simple words: Dummy Variable = Category represented using 0 or 1.

Step 2 – Demonstrate the Concept (Markdown)
Example:

Before:
Color
Red
Blue
Green

after

Color_Blue  Color_Green  Color_Red
    0           0           1
    1           0           0
    0           1           0

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Color": ["Red", "Blue", "Green", "Red"]
})

# Create dummy variables
df = pd.get_dummies(
    df,
    columns=["Color"],
    drop_first=True
)

print(df)
```

## Frequency Encoding
### Step 1 – Understand the Concept (Markdown)
Definition; Frequency Encoding replaces each category with the number of times that category appears in the dataset.

In simple words: Frequency Encoding = Category → How many times it occurs

### Step 2 – Demonstrate the Concept (Markdown)
Example:

Before:

City

Hyderabad → 3
Chennai   → 1
Delhi     → 1

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "City": ["Hyderabad", "Chennai", "Hyderabad",
             "Delhi", "Hyderabad"]
})

# Calculate frequency
freq = df["City"].value_counts()

# Apply frequency encoding
df["City_Frequency"] = df["City"].map(freq)

print(df)
```

## Target Encoding
### Step 1 – Understand the Concept (Markdown)
Definition: Target Encoding replaces each category with the average (mean) of the target variable for that category.

In simple words:Target Encoding = Category → Average target value

### Step 2 – Demonstrate the Concept (Markdown)
Example:Hyderabad → (1+1+0)/3 = 0.67
Chennai   → 0/1 = 0
Delhi     → 0/1 = 0

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "City": ["Hyderabad", "Chennai", "Hyderabad",
             "Delhi", "Hyderabad"],
    "Purchased": [1, 0, 1, 0, 0]
})

# Calculate target mean for each category
target_mean = df.groupby("City")["Purchased"].mean()

# Apply target encoding
df["City_Encoded"] = df["City"].map(target_mean)

print(df)
```

```python
## Handling Unknown Categories
### Step 1 – Understand the Concept (Markdown)
Definition: Unknown Categories are categories that appear in new/test data but were not present during training.

In simple words: Unknown Category = A new category that the encoder has not seen before.

### Step 2 – Demonstrate the Concept (Markdown)
Example:

Training data:
City
Hyderabad
Chennai
Delhi

New data:
City
Hyderabad
Mumbai   ← Unknown category
```

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(
    handle_unknown="ignore"
)

encoder.fit(df[["City"]])

new_data = pd.DataFrame({
    "City": ["Mumbai"]
})

encoded = encoder.transform(
    new_data[["City"]]
)

print(encoded.toarray())
```

```python
## High-Cardinality Categories
### Step 1 – Understand the Concept (Markdown)
Definition: High-cardinality categories are categorical columns that contain a large number of unique values.

In simple words:High Cardinality = Too many unique categories.

### Step 2 – Demonstrate the Concept (Markdown)
Example:
Customer_ID
1001
1002
1003
1004
...
9999

If a column has thousands of unique customer IDs, it has high cardinality.
```

```python
## Step 3 – Implement the Concept (Python)
#Check number of unique categories:
print(df["Customer_ID"].nunique())
```

```python
Nominal       → No order       → One-Hot
Ordinal       → Has order      → Ordinal Encoding
Binary        → 2 categories   → 0 / 1
Label         → Category → Number
One-Hot       → Category → 0/1 columns
Dummy         → Binary columns
Frequency     → Category → Count
Target        → Category → Target Mean
Unknown       → Handle new categories
High Cardinal → Many categories → Frequency/Target Encoding
```

```python
┌────────────────────────────────────────────────────────────┐
│              CATEGORICAL ENCODING – EXAMPLES               │
├────────────────────────────────────────────────────────────┤
│ Nominal                                                     │
│ → Use: One-Hot Encoding                                    │
│ → Example: Color = Red, Blue, Green                        │
│ → No natural order                                         │
│                                                            │
│ Ordinal                                                     │
│ → Use: Ordinal Encoding                                    │
│ → Example: Size = Small < Medium < Large                   │
│ → Has natural order                                        │
│                                                            │
│ Binary                                                      │
│ → Use: 0 / 1 Encoding                                      │
│ → Example: Pass = Yes / No                                 │
│ → Only 2 categories                                        │
│                                                            │
│ Label Encoding                                              │
│ → Use: Label Encoding                                      │
│ → Example: Male = 1, Female = 0                            │
│ → Category → Number                                        │
│                                                            │
│ One-Hot Encoding                                            │
│ → Use: Nominal categories                                  │
│ → Example:                                                   │
│   Red → 1,0,0                                               │
│   Blue → 0,1,0                                              │
│   Green → 0,0,1                                             │
│                                                            │
│ Dummy Variables                                             │
│ → Use: 0/1 representation                                  │
│ → Example:                                                   │
│   Gender_Male = 1/0                                        │
│   Gender_Female = 1/0                                      │
│                                                            │
│ Frequency Encoding                                          │
│ → Use: High-cardinality categories                          │
│ → Example:                                                   │
│   Hyderabad → 100                                          │
│   Chennai → 50                                             │
│                                                            │
│ Target Encoding                                             │
│ → Use: Category + Target available                          │
│ → Example:                                                   │
│   Hyderabad → 0.75                                         │
│   Chennai → 0.40                                           │
│   (Average target value)                                   │
│                                                            │
│ Unknown Categories                                          │
│ → Use: When new category appears                            │
│ → Example: Training = Delhi, Chennai                       │
│            New data = Mumbai → Unknown                      │
│                                                            │
│ High Cardinality                                            │
│ → Use: When there are many unique categories                │
│ → Example: Customer_ID, Product_ID, Email                   │
│ → Usually Frequency / Target Encoding                       │
└────────────────────────────────────────────────────────────┘
```

```python
Color = Red, Blue, Green
        ↓
No order
        ↓
One-Hot Encoding ✅

Size = Small, Medium, Large
        ↓
Has order
        ↓
Ordinal Encoding ✅

Pass = Yes, No
        ↓
Only 2 categories
        ↓
0 / 1 Encoding ✅

Customer_ID = 1001, 1002, 1003, ... 100000
        ↓
Many unique values
        ↓
Frequency / Target Encoding ✅
```

```python
| **Column**    | **Type**             | **When to Use**          | **Encoding**     | **Example**                      |
| ------------- | -------------------- | ------------------------ | ---------------- | -------------------------------- |
| `Gender`      | Binary               | Only 2 categories        | 0/1 Encoding     | Male → 1, Female → 0             |
| `Color`       | Nominal              | No order                 | One-Hot          | Red → `[1,0,0]`                  |
| `City`        | Nominal              | Few categories, no order | One-Hot          | Hyderabad → `[1,0,0]`            |
| `Education`   | Ordinal              | Has natural order        | Ordinal          | School → 0, Graduate → 1, PG → 2 |
| `Size`        | Ordinal              | Has ranking              | Ordinal          | Small → 0, Medium → 1, Large → 2 |
| `Pass_Status` | Binary               | Yes/No                   | 0/1              | Yes → 1, No → 0                  |
| `Department`  | Nominal              | Few categories           | One-Hot          | IT, HR, Sales                    |
| `Customer_ID` | High Cardinality     | Many unique values       | Frequency        | ID → Frequency                   |
| `Product_ID`  | High Cardinality     | Many unique values       | Frequency/Target | Product → Count/Mean             |
| `City`        | High Cardinality     | Many unique cities       | Frequency/Target | Hyderabad → 0.75                 |
| `Category`    | Categorical + Target | Target available         | Target Encoding  | Category → Target Mean           |
```

```python
┌─────────────────────────────────────────┐
│ Gender (Yes/No)       → 0 / 1          │
│ Color (No Order)      → One-Hot        │
│ Size (Has Order)      → Ordinal        │
│ Many Unique Values    → Frequency      │
│ Target Available      → Target         │
└─────────────────────────────────────────┘
```

# Data Preprocessing Fundamentals

## What is Data Preprocessing?

### Step 1 – Understand the Concept (Markdown)
Definition: Data preprocessing is the process of cleaning, transforming, and preparing raw data so that it can be used effectively for data analysis and machine learning models.

In simple words: Raw data is often messy, incomplete, or inconsistent. Data preprocessing converts this raw data into a clean and useful format.

Common preprocessing tasks include: Handling missing values, Removing duplicate data, Handling outliers, Converting categorical data into numerical data

### Step 2 – Demonstrate the Concept (Markdown

| Name  | Age | Gender | Salary |
| ----- | --: | ------ | -----: |
| Ravi  |  25 | Male   |  30000 |
| Priya |     | Female |  40000 |
| Ravi  |  25 | Male   |  30000 |
| John  | 200 | Male   |  35000 |

Age is missing for Priya.
Ravi's record is duplicated.
John's age 200 is an outlier/invalid value.
Gender contains text values that may need encoding.

LabelEncoder → converts categorical/text values into numbers.
StandardScaler → scales numerical values so they have a similar scale.

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler

data = {
    "Name": ["Ravi", "Priya", "Ravi", "John"],
    "Age" : [20,30,20,None],
    "Gender": ["Male", "Female", "Male", "Male"],
    "Salary": [30000, 40000, 30000, 35000]
}
df = pd.DataFrame(data)


# 1. Handle missing value
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)

df = df.drop_duplicates()

print("after removing duplicates")
print(df)

# "Find all ages greater than 100 and replace them with the median age."
df.loc[df["Age"] > 100, "Age"] = df["Age"].median

# 4. Encode categorical data
encoder = LabelEncoder()
df["Gender"] = encoder.fit_transform(df["Gender"])

print("categorical encoding")
print(df)


#Numerical scaling
scaler = StandardScaler()
df[["Age", "Salary"]] = scaler.fit_transform(df[["Age", "Salary"]])
print(df)

print("\nPreprocessed Data:")
print(df)
```

## Data Quality

### Step 1 – Understand the Concept (Markdown)
Definition: Data quality refers to how accurate, complete, consistent, reliable, and useful the data is for a particular purpose.

In simple words: Good quality data = Correct + Complete + Consistent + Reliable data

Before doing data analysis or machine learning, we need to check whether the data is of good quality.

Main dimensions of Data Quality:
Accuracy – Is the data correct?
Completeness – Are required values present?
Consistency – Is the data consistent across records?
Validity – Does the data follow the expected rules?
Uniqueness – Are there duplicate records?
Timeliness – Is the data current and up to date?

## Data Leakage
### Step 1 – Understand the Concept (Markdown)
Definition: Data leakage occurs when information from outside the training data, especially the test/future data, unintentionally influences the model during training.

In simple words: Data Leakage = The model gets information it should not have during training.

This can make the model's performance look better than it really is.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example; Suppose we want to predict whether a patient has cancer.

If we use a feature like:

Treatment Given;and the treatment was decided after knowing the cancer diagnosis, then using it to predict the diagnosis causes data leakage.

The model gets information about the answer in advance. ❌

Data Leakage = Training data gets information it should not have → Unrealistically high model performance.

## Preprocessing Pipeline
### Step 1 – Understand the Concept (Markdown)
Definition: A preprocessing pipeline is a sequence of preprocessing steps that are automatically applied to data in a specific order before giving it to a machine learning model.

In simple words: Pipeline = Multiple preprocessing steps combined into one workflow.
### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Instead of doing:

Missing Values → Encoding → Scaling → Model

separately, we create one pipeline:

Raw Data
   ↓
Imputation
   ↓
Encoding
   ↓
Scaling
   ↓
Machine Learning Model

### Step 3 – Implement the Concept (Python)

```python
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler

pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

X_train_processed = pipeline.fit_transform(X_train)
X_test_processed = pipeline.transform(X_test)
```

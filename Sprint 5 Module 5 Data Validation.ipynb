# Data Validation

## What is Data Validation?
### Step 1 – Understand the Concept (Markdown)
Definition: Data Validation is the process of checking whether data is correct, valid, and follows predefined rules.

In simple words: Data Validation = Checking whether data is correct and acceptable.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
    
Age → 25       ✅ Valid
Age → -5       ❌ Invalid
Email → a@gmail.com  ✅ Valid
Email → abc    ❌ Invalid

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

data = {
    "Name" : ["mahi", "raju"],
    "age"  : [20,10],
    "salary": [20000,-50000],
    "gender" : ["Male", "Female"]
}
df = pd.DataFrame(data)

df
```

## Range Validation
### Step 1 – Understand the Concept (Markdown)
Definition: Range Validation checks whether a value falls within a specified minimum and maximum range.

In simple words:Range Validation = Check whether a value is within the allowed range.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Age = 25   ✅ Valid (0–100)
Age = 150  ❌ Invalid
Age = -5   ❌ Invalid

### Step 3 – Implement the Concept (Python)

```python
# Keep Age between 0 and 100

df = df[(df["age"] >= 18 ) & (df["age"] <= 100)]
df
```

## Type Validation

In simple words: Type Validation = Check whether data is in the correct type.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Age = 25       ✅ Integer

## Category Validation
### Step 1 – Understand the Concept (Markdown)
Definition: Category Validation checks whether values belong to a predefined list of allowed categories.

In simple words: Category Validation = Check whether the value is an allowed category.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Allowed categories:

IT, HR, Sales
IT       ✅ Valid
HR       ✅ Valid
Finance  ❌ Invalid

### Step 3 – Implement the Concept (Python)

```python
allowed = ["IT", "HR", "Sales"]

df["Valid"] = df["Department"].isin(allowed)

print(df)
```

## Null Validation
### Step 1 – Understand the Concept (Markdown)
Definition:Null Validation checks whether required fields contain missing or null values.

In simple words: Null Validation = Check whether required data is present.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Name   → "Mani"   ✅ Valid
Age    → 25       ✅ Valid
Email  → NaN      ❌ Invalid

### Step 3 – Implement the Concept (Python)

```python
# Check null values
print(df.isnull().sum())

# Check if Age has no null values
print(df["age"].notnull())
```

```python
## Business Rule Validation
### Step 1 – Understand the Concept (Markdown)
Definition: Business Rule Validation checks whether data follows the specific rules defined by the business.

In simple words: Business Rule Validation = Check whether data follows business requirements.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Business rule:

Age must be 18 or above.
Salary must be greater than 0.
Age = 25      ✅ Valid
Age = 15      ❌ Invalid
Salary = 30000 ✅ Valid
Salary = -500  ❌ Invalid

### Step 3 – Implement the Concept (Python)
```

## Constraint Validation
### Step 1 – Understand the Concept (Markdown)
Definition:Constraint Validation checks whether data follows predefined restrictions or conditions.

In simple words: Constraint Validation = Check whether data follows required rules or limits.

### Step 2 – Demonstrate the Concept (Markdown)
Age must be ≥ 18
Salary must be > 0
Email must not be empty

### Step 3 – Implement the Concept (Python)

```python
# Check constraints
df["Valid"] = (
    (df["age"] >= 18) &
    (df["salary"] > 0)
)

print(df)
```

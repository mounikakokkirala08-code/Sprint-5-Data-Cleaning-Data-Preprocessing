## Bivariate Analysis

## 1. Numerical vs Numerical ; 



### Step 1 – Understand the Concept
Own Words

Here we study the relationship between two numerical variables.

Your dataset examples
Area_sqft       ↔ Price_lakhs
Bedrooms        ↔ Price_lakhs
Bathrooms       ↔ Price_lakhs
Age_years       ↔ Price_lakhs
Location_Score  ↔ Price_lakhs
Parking_Spaces  ↔ Price_lakhs

The most important relationship for your project is:

Area_sqft ↔ Price_lakhs
### Step 2 – Real-World Analogy

Suppose:

House A → 1000 sqft → ₹50 lakhs
House B → 2000 sqft → ₹100 lakhs
House C → 3000 sqft → ₹180 lakhs

We want to know:

As area increases, does price also increase?

### Step 3 – Implement the Concept


```python
import pandas as pd
df = pd.read_csv("house_price_prediction_cleaned.csv")
```

```python
#Scatter Plot
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Area_sqft",
    y="Price_lakhs"
)

plt.title("Area vs House Price")
plt.xlabel("Area (sqft)")
plt.ylabel("Price (Lakhs)")

plt.show()
```

```python
### 2. Numerical vs Categorical
### Step 1 – Understand the Concept

Here we study a numerical variable across different categories.

Your dataset examples
Price_lakhs      ↔ Property_Type
Price_lakhs      ↔ Furnishing_Status
Price_lakhs      ↔ Construction_Status
Price_lakhs      ↔ Locality_Type
Area_sqft        ↔ Property_Type

For your project, an important example is:

Property_Type ↔ Price_lakhs
### Step 2 – Real-World Analogy

We want to know:

Do Apartment, Independent House, and Villa properties have different house prices?

For example:

Apartment
→ What is the typical price?

Independent House
→ What is the typical price?

Villa
→ What is the typical price?
### Step 3 – Implement the Concept
Group Statistics

Interpretation

Look for:

Median
Spread
Range
Potential outliers
```

```python
df.groupby("Property_Type")["Price_lakhs"].mean()
```

```python
#Mean, Median, Minimum and Maximum
df.groupby("Property_Type")["Price_lakhs"].agg([
    "mean",
    "median",
    "min",
    "max"
])
# Box Plot

#A box plot is very useful for Numerical vs Categorical.

plt.figure(figsize=(8, 5))

sns.boxplot(
    data=df,
    x="Property_Type",
    y="Price_lakhs"
)

plt.title("House Price by Property Type")
plt.xlabel("Property Type")
plt.ylabel("Price (Lakhs)")

plt.xticks(rotation=30)

plt.show()
```

## Grouped Bar Chart	 Categorical vs Categorical


```python
plt.figure(figsize=(9, 5))

sns.countplot(data=df, x = "Property_Type", hue = "Furnishing_Status" )

plt.title("Grouped Bar Chart")

plt.xlabel("Property_Type")

plt.ylabel("Furnishing_Status")

plt.show()
```

```python
plt.figure(figsize = (9,3))

sns.countplot(data=df, x = "Construction_Status", hue = "Locality_Type")

plt.title("Grouped Bar Chart")

plt.xlabel("Construction_Status")

plt.ylabel("Locality_Type")
```

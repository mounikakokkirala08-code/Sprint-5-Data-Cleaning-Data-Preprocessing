# Feature Scaling

## Feature Scaling
### Step 1 – Understand the Concept (Markdown)

Definition: Feature Scaling is a data preprocessing technique that brings numerical features to a similar scale.

In simple words: It prevents large-value features from dominating small-value features.

### Step 2 – Demonstrate the Concept (Markdown)

Simple example: Before scaling:

Age      = 25
Salary   = 50000   After scaling:   Age      = 0.25, Salary   = 0.50

Both features are now on a comparable scale.

Common techniques:

StandardScaler
MinMaxScaler
RobustScaler

## Feature Scaling – Types for Numerical Data
### Step 1 – Understand the Concept (Markdown)

Definition: Numerical feature scaling has different techniques to bring numerical values to a suitable scale.

In simple words: We change the range/distribution of numerical values so ML algorithms can work better.

### Step 2 – Demonstrate the Concept (Markdown)

Main types:

1.Standardization (StandardScaler)
Mean = 0, Standard Deviation = 1
Commonly used in ML.
    
2 Min-Max Scaling (MinMaxScaler)
Converts values usually between 0 and 1.
Useful for Neural Networks and distance-based algorithms.
    
3.Robust Scaling (RobustScaler)
Uses Median and IQR.
Good when data contains outliers.
    
4.Max Absolute Scaling (MaxAbsScaler)   
Scales values between -1 and 1.
Useful for sparse data.
    
### Step 3 – Implement the Concept (Python)

```python
from sklearn.preprocessing import (
    StandardScaler,
    MinMaxScaler,
    RobustScaler,
    MaxAbsScaler
)

# Standardization
StandardScaler()

# Min-Max Scaling
MinMaxScaler()

# Robust Scaling
RobustScaler()

# Max Absolute Scaling
MaxAbsScaler()
```

## Normalization

### Step 1 – Understand the Concept (Markdown)

Definition: Normalization is a feature scaling technique that converts numerical values into a fixed range, usually 0 to 1.

In simple words: It makes different numerical features have a similar scale.

### Step 2 – Demonstrate the Concept (Markdown)

Simple example: Original values: 10, 20, 30, 40, 50

After Min-Max Normalization: 0, 0.25, 0.5, 0.75, 1

Formula:

X' = (X - Xmin) / (Xmax - Xmin)
Normalization → usually converts values to 0–1 range → MinMaxScaler

### Step 3 – Implement the Concept (Python)

```python
from sklearn.preprocessing import MinMaxScaler

data = [[10], [20], [30], [40], [50]]

scaler = MinMaxScaler()

normalised_data = scaler.fit_transform(data)

normalised_data

#Normalization specifically means Min-Max scaling (0–1).
```

## Standardization
### Step 1 – Understand the Concept (Markdown)

Definition: Standardization is a feature scaling technique that transforms numerical data so that:

Mean = 0
Standard Deviation = 1

In simple words: It tells us how far each value is from the average.
### Step 2 – Demonstrate the Concept (Markdown)

Formula:

Z = (X - Mean) / Standard Deviation

Example:

Data: 10, 20, 30
Mean = 20

After standardization, the values are approximately:

-1.22, 0, 1.22
### Step 3 – Implement the Concept (Python)


```python
from sklearn.preprocessing import StandardScaler


data = [[10], [20], [30]]


scaler = StandardScaler()


standardized = scaler.fit_transform(data)


print(standardized)
```

## Min-Max Scaling
### Step 1 – Understand the Concept (Markdown)

Definition: Min-Max Scaling is a feature scaling technique that converts numerical values into a fixed range, usually 0 to 1.

In simple words ; It brings the smallest value to 0 and the largest value to 1.

Step 2 – Demonstrate the Concept (Markdown)

Formula:

X' = (X - Xmin) / (Xmax - Xmin)

Example:

Original:     10   20   30   40   50
Scaled:        0  0.25  0.5  0.75   1

Range: Usually 0 to 1.

### Step 3 – Implement the Concept (Python)

```python
from sklearn.preprocessing import MinMaxScaler

data = [[10], [20], [30], [40], [50]]

scaler = MinMaxScaler()

result = scaler.fit_transform(data)

print(result)
```

## StandardScaler
### 1 – Understand the Concept (Markdown)

Definition: StandardScaler is a Scikit-learn preprocessing tool used to standardize numerical data.

In simple words:It converts numerical features so that:

Mean ≈ 0
Standard Deviation ≈ 1
Range → No fixed range
### Step 2 – Demonstrate the Concept (Markdown)

Example:

Original:       10   20   30
Standardized: -1.22   0   1.22

It uses:

Z = (X - Mean) / Standard Deviation

Unlike MinMaxScaler, it does not restrict values to 0–1.

### Step 3 – Implement the Concept (Python)

```python
from sklearn.preprocessing import StandardScaler

data = [[10], [20], [30]]

scaler = StandardScaler()

result = scaler.fit_transform(data)

print(result)
```

```python
Standardization = Technique
StandardScaler = Python tool
Mean = 0, SD = 1, No fixed range
```

## RobustScaler
### Step 1 – Understand the Concept (Markdown)

Definition: RobustScaler is a Scikit-learn scaling technique that uses the median and IQR (Interquartile Range) to scale numerical data.

In simple words:Tt is useful when your data contains outliers.

### Step 2 – Demonstrate the Concept (Markdown)

Formula: Scaled = (X - Median) / IQR

Example: Data: 10, 20, 30, 40, 1000
                         ↑
                      Outlier

RobustScaler is less affected by 1000 because it uses Median and IQR instead of Mean and Standard Deviation.

Range: ❌ No fixed range.

### Step 3 – Implement the Concept (Python)

```python
from sklearn.preprocessing import RobustScaler

data = [[10], [20], [30], [40], [1000]]

scaler = RobustScaler()

result = scaler.fit_transform(data)

print(result)
```

## MaxAbsScaler
### Step 1 – Understand the Concept (Markdown)

Definition: MaxAbsScaler is a Scikit-learn scaling technique that scales numerical values using the maximum absolute value of each feature.

In simple words: It converts values so that they are generally within -1 to +1, without changing the sign of the data.

### Step 2 – Demonstrate the Concept (Markdown)

Formula: Scaled = X / Maximum Absolute Value

Example:

Original:   -10   -5    0    5    10
Scaled:     -1   -0.5   0   0.5    1

Range: Usually -1 to +1.

It is especially useful for sparse data because it does not center the data around zero.

### Step 3 – Implement the Concept (Python)

```python
from sklearn.preprocessing import MaxAbsScaler

data = [[-10], [-5], [0], [5], [10]]

scaler = MaxAbsScaler()

result = scaler.fit_transform(data)

print(result)
```

## Scale-Sensitive Algorithms
### Step 1 – Understand the Concept (Markdown)

Definition: Scale-sensitive algorithms are ML algorithms whose performance is affected by the scale or magnitude of numerical features.

In simple words: f one feature has values like 10 and another has values like 50,000, the larger feature can dominate the model.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Age       → 20, 30, 40
Salary    → 20,000, 30,000, 40,000

Without scaling, Salary has much larger values.

After scaling:

Age       → -1, 0, 1
Salary    → -1, 0, 1

Now both features have a comparable scale.

Common scale-sensitive algorithms:

KNN
K-Means
SVM
Logistic Regression
Linear Regression (especially gradient-based optimization)
Neural Networks
PCA

## Scale-Insensitive Algorithms
### Step 1 – Understand the Concept (Markdown)

Definition: scale-insensitive algorithms are ML algorithms that are generally not affected by the numerical scale of features.

In simple words: They can work with features having different ranges without requiring feature scaling.

### Step 2 – Demonstrate the Concept (Markdown)

Example:

Age      → 20, 30, 40
Salary   → 20,000, 30,000, 40,000

For tree-based models, scaling is usually not required because they make decisions using feature thresholds.

Common scale-insensitive algorithms:

Decision Tree
Random Forest
XGBoost
LightGBM
CatBoost

```python
| Technique                           | Python Tool      | Range          | Best Use              |
| ----------------------------------- | ---------------- | -------------- | --------------------- |
| **Normalization / Min-Max Scaling** | `MinMaxScaler`   | 0 to 1         | Distance-based models |
| **Standardization**                 | `StandardScaler` | No fixed range | General ML            |
| **Robust Scaling**                  | `RobustScaler`   | No fixed range | Data with outliers    |
| **Max Absolute Scaling**            | `MaxAbsScaler`   | -1 to +1       | Sparse data           |
```

```python
Feature Scaling
│
├── Min-Max Scaling → MinMaxScaler → 0 to 1
├── Standardization → StandardScaler → Mean 0, SD 1
├── Robust Scaling → RobustScaler → Median + IQR
└── MaxAbs Scaling → MaxAbsScaler → -1 to +1
```

```python
Numerical data → Scaling
Categorical data → Encoding
KNN/K-Means/SVM/NN → Scaling important
Tree models → Scaling usually not required
```

```python
Scale-Sensitive Algorithms → Scaling usually required

KNN
K-Means
SVM
Logistic Regression
Neural Networks
PCA

Scale-Insensitive Algorithms → Scaling usually not required

Decision Tree
Random Forest
XGBoost
LightGBM
CatBoost
```

Key Point

Outliers strongly affect MinMaxScaler, StandardScaler, and MaxAbsScaler.

RobustScaler is the best choice when significant outliers are present.

```python
| Technique           | Tool             | Range          | Outlier Effect    |
| ------------------- | ---------------- | -------------- | ----------------- |
| **Min-Max Scaling** | `MinMaxScaler`   | 0 to 1         | ❌ Highly affected |
| **Standardization** | `StandardScaler` | No fixed range | ⚠️ Affected       |
| **Robust Scaling**  | `RobustScaler`   | No fixed range | ✅ Less affected   |
| **Max Abs Scaling** | `MaxAbsScaler`   | -1 to 1        | ❌ Affected        |
```

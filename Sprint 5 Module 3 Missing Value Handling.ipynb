# Missing Value Handling

```python
## What are Missing Values?
### Step 1 – Understand the Concept (Markdown)
Definition: Missing values are values that are not available or not recorded in a dataset.

    In simple words: Missing Value = Data is empty or unavailable. Common representations:

NaN
None
NULL
Empty cells
    
Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Name	Age	Salary
Mani	25	30000
Roja	NaN	40000
Mahi	30	NaN

Here, Roja's Age and Mahi's Salary are missing values.

Step 3 – Implement the Concept (Python)
```

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    "Name": ["Mani", "Roja", "Mahi"],
    "Age": [25, np.nan, 30],
    "Salary": [30000, 40000, np.nan]
})

# Check missing values
print(df.isnull())
```

## Mean Imputation

### Step 1 – Understand the Concept (Markdown)
Definition: Mean imputation is a technique for handling missing numerical values by replacing them with the mean (average) of that column.

In simple words: Missing value → Replace it with the average value.

### Step 2 – Demonstrate the Concept (Markdown)

Age
20
30
NaN

(20 + 30 + 40) / 3 = 30

```python
### Step 3 – Implement the Concept (Python)
import pandas as pd

df = pd.DataFrame({
    "Age": [20, 30, None, 40]
})

# Replace missing value with mean
df["Age"] = df["Age"].fillna(df["Age"].mean())

print(df)
```

## Median Imputation
### Step 1 – Understand the Concept (Markdown)

In simple words: Missing value → Replace it with the middle value.

```python
import pandas as pd


df = pd.DataFrame({
    "Age": [20, 30, None, 40, 50]
})


# Replace missing value with median
df["Age"] = df["Age"].fillna(df["Age"].median())


print(df)
```

```python
## Mode Imputation
### Step 1 – Understand the Concept (Markdown)
Definition: Mode imputation is a technique for handling missing values by replacing them with the most frequently occurring value in a column.

In simple words: Missing value → Replace it with the most common value.

Step 2 – Demonstrate the Concept (Markdown)
Simple example:
Gender
Male
Female
Male
NaN
Male

The most frequent value is:

Male
```

### Step 3 – Implement the Concept (Python)

```python
import pandas as pd

df = pd.DataFrame({
    "Gender": ["Male", "Female", "Male", None, "Male"]
})

# Replace missing value with mode
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])

print(df)
```

## Forward Fill
### Step 1 – Understand the Concept (Markdown)
Definition: Forward Fill (ffill) is a technique that replaces a missing value with the previous available value in the column.

In simple words:

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [20, None, 30, None, 40]
})

# Forward fill
df["Age"] = df["Age"].ffill()

print(df)
```

```python
## Backward Fill
### Step 1 – Understand the Concept (Markdown)
Definition: Backward Fill (bfill) is a technique that replaces a missing value with the next available value in the column.

In simple words: Missing value → Take the value from below.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Before:

Age
20
NaN
30
NaN
40

After Backward Fill:

Age
20
30
30
40
40
```

```python
### Step 3 – Implement the Concept (Python)

import pandas as pd

df = pd.DataFrame({
    "Age": [20, None, 30, None, 40]
})


# Backward fill
df["Age"] = df["Age"].bfill()


print(df)
```

## Group-Based Imputation
### Step 1 – Understand the Concept (Markdown)
Definition: Group-Based Imputation means filling missing values using a statistic such as mean, median, or mode calculated within a specific group.

In simple words: Missing value → Fill using the value from its own group.

### Step 2 – Demonstrate the Concept (Markdown)

| Department | Salary |
| ---------- | -----: |
| IT         | 50,000 |
| IT         |    NaN |
| HR         | 30,000 |
| HR         | 40,000 |

### Step 3 – Implement the Concept (Python)



```python
import pandas as pd

df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR"],
    "Salary": [50000, None, 30000, 40000]
})

# Fill missing salary using department-wise median
df["Salary"] = df["Salary"].fillna(
    df.groupby("Department")["Salary"].transform("median")
)

print(df)
```

## KNN Imputation
### Step 1 – Understand the Concept (Markdown)
Definition: KNN Imputation uses the values of similar data points (nearest neighbors) to estimate and fill missing values.

In simple words: KNN Imputation = Find similar rows → Use their values to fill the missing value.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Suppose a person's Age is missing.

KNN looks for other people with similar features, such as:

Height, Weight, Salary, etc.

If the nearest neighbors have ages:

25, 27, 26

KNN can estimate the missing age using these nearby values
    
### Step 3 – Implement the Concept (Python)

```python
import pandas as pd
from sklearn.impute import KNNImputer

df = pd.DataFrame({
    "Age": [25, 30, None, 40],
    "Salary": [30000, 40000, 35000, 50000]})

# Create KNN imputer
imputer = KNNImputer(n_neighbors=2)

# Fill missing values
df_imputed = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)

print(df_imputed)
```

## Iterative Imputation
### Step 1 – Understand the Concept (Markdown)
Definition: Iterative Imputation fills missing values by using other features to predict the missing value, repeating the process until the estimates become stable.

In simple words: Iterative Imputation = Predict missing values using other columns, repeatedly.

### Step 2 – Demonstrate the Concept (Markdown)
Simple example:

Suppose:

Age	Salary
25	30000
30	NaN
40	50000

Step 3 – Implement the Concept (Python)

```python
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer

df = pd.DataFrame({
    "Age": [25, 30, 40],
    "Salary": [30000, None, 50000]
})

# Create Iterative Imputer
imputer = IterativeImputer(random_state=42)

# Fill missing values
df_imputed = pd.DataFrame(
    imputer.fit_transform(df),
    columns=df.columns
)

print(df_imputed)
```

```python
| Technique         | Best Used For                        | Main Advantage                    | Main Limitation                |
| ----------------- | ------------------------------------ | --------------------------------- | ------------------------------ |
| **Mean**          | Numerical, normal data               | Simple & fast                     | Sensitive to outliers          |
| **Median**        | Numerical, skewed data               | Robust to outliers                | Reduces variability            |
| **Mode**          | Categorical data                     | Very simple                       | Can increase majority bias     |
| **Constant**      | Meaningful default/missing indicator | Easy & explicit                   | Can create artificial patterns |
| **Forward Fill**  | Time-series                          | Simple & preserves previous value | Can propagate errors           |
| **Backward Fill** | Time-series                          | Simple & uses next value          | Can propagate future values    |
| **Interpolation** | Ordered numerical data               | Uses surrounding values           | Assumes a pattern              |
| **Group-Based**   | Group-dependent data                 | Preserves group patterns          | Small groups can be unreliable |
| **KNN**           | Related features                     | Uses similar observations         | Computationally expensive      |
| **Iterative**     | Strong feature relationships         | Uses multiple features            | Complex and slower             |
```

```python
Numerical + Normal data       → Mean
Numerical + Outliers/Skewed   → Median
Categorical                   → Mode
Meaningful fixed category     → Constant
Time-series                   → Forward/Backward Fill
Ordered numerical data        → Interpolation
Different groups              → Group-Based
Similar observations          → KNN
Strong feature relationships  → Iterative
```
